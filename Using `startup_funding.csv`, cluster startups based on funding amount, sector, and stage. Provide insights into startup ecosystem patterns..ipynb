```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("startup_funding.csv")
```

```python
le_sector = LabelEncoder()
le_stage = LabelEncoder()
```

```python
df["Sector_enc"] = le_sector.fit_transform(df["Sector"])
df["Stage_enc"] = le_stage.fit_transform(df["Stage"])
```

```python
X = df[["FundingAmount_CrINR", "Sector_enc", "Stage_enc"]]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
df["Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
print(df[["StartupName", "Sector", "Stage", "FundingAmount_CrINR", "Cluster"]].head())
```

```python
print("\nCluster wise average funding:")
print(df.groupby("Cluster")["FundingAmount_CrINR"].mean())
```

#### 4 clustersum avarde averagum

```python
print("\nCluster Sizes:")
print(df["Cluster"].value_counts())
```

```python
print("\nAverage Funding per Cluster:")
print(df.groupby("Cluster")["FundingAmount_CrINR"].mean())
```

```python
print("\nDominant Stage per Cluster:")
print(df.groupby("Cluster")["Stage"].agg(lambda x: x.value_counts().index[0]))
```

```python
print("\nFunding Range per Cluster:")
print(df.groupby("Cluster")["FundingAmount_CrINR"].agg(["min", "max"]))
```
